```python
from sklearn.model_selection import train_test_split #we need 2 parameter i.data , ii. label(outcome of data in csv)
```

```python
x=df.iloc[:,df.columns!='Outcome'] #data
y=df.iloc[:,df.columns=='Outcome'] #label
```

```python
print(x)
```

```python
xtrain, xtext, ytrain, ytest = train_test_split(x,y,test_size=0.2) #20%test data
```

```python
xtrain.head() 
```

```python
ytrain.head()#.shape
```

ALGORITHM

```python
from sklearn.ensemble import RandomForestClassifier #RandomForestReggersion
```

```python
model= RandomForestClassifier() #to giving new test data
```

```python
model.fit(xtrain,ytrain.values.ravel()) #used to train algorithm in both ml , dl
```

```python

predict_output=model.predict(xtext) #to test the algorithm
print(predict_output) 
```

```python
#comparing both csv and algorithm data
from sklearn.metrics import accuracy_score
```

```python
acc=accuracy_score(predict_output,ytest)  #to know the accurasy
```

```python
print(' RF score is :', acc)
```
