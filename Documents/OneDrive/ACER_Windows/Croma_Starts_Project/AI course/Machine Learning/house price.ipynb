```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.datasets
from sklearn.model_selection import train_test_split
from sklearn.datasets import fetch_california_housing
from xgboost import XGBRegressor
from sklearn import metrics
```

```python
   hp = fetch_california_housing()
   hp
```

```python
#target 14th column represent price (label)
#here we changing the structure to understand the imporper structure
```

```python
hpdf=pd.DataFrame(hp.data , columns =hp.feature_names)
hpdf
```

```python
hpdf.head() #first 5
```

```python
#include target(house price) =>label
hpdf ['Price'] = hp.target #adding target as price in column
hpdf.head()
```

```python
X = hpdf.drop(['Price'],axis=1) #for row axis=0
X
```

```python
Y= hpdf['Price']
Y
```

# Spliting Train and Test

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y, test_size=0.2,random_state=2)  #tran x and test X will store in trained and tested y
```

```python

X_train.shape
```

```python
X_test.shape
```

```python
X.shape
```

# MODEL TRAINING

# XGBoost Regression

```python
#loading the model
model= XGBRegressor()
```

```python
#training the model with X,Y_training set
model.fit(X_train,Y_train) #fit is the code
```

prediction of the trained data

```python
#error predction of train data
#accuracy on predction on trained data set
predction = model.predict(X_train)
predction
```

# EVALUATION

```python

# R square error
score_1= metrics.r2_score(Y_train,predction)

#mean square error
score_2= metrics.mean_absolute_error(Y_train,predction)
print("R square error :",score_1)
print("mean square erroe :",score_2)
```

```python
#error predction of test data
predctiont = model.predict(X_test)
predctiont

# R square error
score_1_test = metrics.r2_score(Y_test,predctiont)

#mean square error
score_2_test = metrics.mean_absolute_error(Y_test,predctiont)
print("R square error :",score_1_test)
print("mean square erroe :",score_2_test)
```

```python
#ploting the error values
plt.scatter(Y_train,predction)
plt.xlabel("Actual preice")
plt.ylabel("Predicted price")
plt.title("Actual VS Predicted")
plt.show()
```
